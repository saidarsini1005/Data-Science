```python
import pandas as pd

# Reading the online CSV dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

# Writing dataset into a CSV file
df.to_csv("iris_output.csv", index=False)

print("The dataset has been successfully written to iris_output.csv")
```

```python
import pandas as pd

# 1. Online Tab-Delimited (TSV) Dataset URL
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"

# 2. Reading a tab-delimited file using pd.read_table()
df = pd.read_table(url)

# 3. Displaying the first five rows
print("First five rows of the dataset:")
print(df.head())

# 4. Displaying dataset summary (shape and info)
print("\nDataset Shape:", df.shape)

# 5. Writing the DataFrame into a new CSV file (excluding index)
df.to_csv("chipotle_summary.csv", index=False)
print("\nData successfully saved to 'chipotle_summary.csv'")
```

```python
import pandas as pd

# Define student data dictionary
student_data = {
    "Roll_No": [101, 102, 103, 104],
    "Name": ["Anusha", "Sabitha", "Chaitra", "Deepika"],
    "Department": ["IT", "IT", "CSE", "ECE"],
    "Percentage": [89, 92, 88, 85]
}

# Create DataFrame
df = pd.DataFrame(student_data)

# Writing data into a tab-delimited file
df.to_csv("student_output.txt", sep="\t", index=False)

print("Data successfully written to student_output.txt")
```

```python
# Creating the sample 'students.txt' file programmatically
data = """Roll-No\tName\tDepartment\tPercentage
101\tAnusha\tIT\t89
102\tSabitha\tIT\t92
103\tCharitha\tCSE\t88
104\tDeepika\tGCE\t85"""

with open("students.txt", "w") as f:
    f.write(data)
```
